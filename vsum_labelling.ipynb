```python
import os 
import cv2
import numpy as np
```

```python
def get_video_fps(video_path):
    probed_video = cv2.VideoCapture(video_path)
    fps = probed_video.get(cv2.CAP_PROP_FPS)
    probed_video.release()
    return fps

def get_video_length(video_path):
    probed_video = cv2.VideoCapture(video_path)
    fps = probed_video.get(cv2.CAP_PROP_FRAME_COUNT)
    probed_video.release()
    return fps
```

```python
# First we list out all the GT and the videos. Then take central frame before and after the frame and label that as the GT
base_video_dir = "C:/Datasets/VSUMM"
valid_extensions = (".mp4", ".avi", ".flv")
youtube_database = os.listdir(os.path.join(base_video_dir,'new_database'))
youtube_database = [x for x in youtube_database if x.lower().endswith(valid_extensions)]
video_names = [x.split('.')[0] for x in youtube_database]
youtube_summary_dir = 'newUserSummary'
youtube_video_dir = "new_database"
joined_youtube_paths = [os.path.join(base_video_dir,youtube_video_dir,x) for x in youtube_database if x.lower().endswith(valid_extensions)]
videos_fps = [get_video_fps(video_path) for video_path in joined_youtube_paths]
videos_n_frames = [get_video_length(video_path) for video_path in joined_youtube_paths]
```

```python
video_num_to_youtube_map = {}
youtube_metadata_dict = {} # should have shot_bounds, n_frames, user_summary
shot_length = 5
original_fps_cases = []
for i,(video_name,video_path,fps,n_frames) in enumerate(zip(video_names,joined_youtube_paths
                                                          ,videos_fps,videos_n_frames)):
    # Create a set of zeros for a given video
    j+=1 
    video_num = f"video_{i+1}"
    base_path = os.path.join(base_video_dir,youtube_summary_dir,video_name)
    num_users = os.listdir(base_path)
    user_summary = np.zeros((len(num_users),int(n_frames)))
    #
    shot_length_frames = int(fps)*shot_length
    shot_boundaries = np.array([[start, min(start+shot_length_frames,int(n_frames))] for start in range(0,int(n_frames),shot_length_frames)])
    for i, user in enumerate(num_users):
        # Find the list of frames
        if user.endswith('.db'):
            continue
        base_user_path = os.path.join(base_path,user)
        list_of_gts = os.listdir(base_user_path)
        print(list_of_gts)
        sampled_indices = [int(gt.split('.')[0][5:]) for gt in list_of_gts if not gt.endswith('.db')]
        print(sampled_indices)
        max_val = max(sampled_indices)
        mul_val = 1
        if max_val/n_frames < 0.6:
            mul_val = int(fps)
        else:
            if video_name not in original_fps_cases:
                original_fps_cases.append(video_name)
        for sampled_index in sampled_indices:
            range_lower = int(max(0,sampled_index*mul_val- fps//2))
            range_upper = int(min(sampled_index*mul_val + fps//2,n_frames))
            print("Ranges: ",range_lower,range_upper)
            user_summary[i,range_lower:range_upper] = 1
    video_num_to_youtube_map[video_name] = video_num
    youtube_metadata_dict[video_num] = {'user_summary':user_summary,
                                        'n_frames':n_frames,
                                    'shot_bounds': shot_boundaries}
```
